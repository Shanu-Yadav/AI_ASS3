```python
def minmax1(oppo_move,my_move):
  a=[]
  b=[]
  D={}
  for i in range(len(oppo_move)):
    a.append(str(oppo_move[i])[:])

  for j in range(len(my_move)):
    b.append(str(my_move[j])[:])
  # print(b)
  # print("a,b=",a,b)
  for i in range(len(b)):
    my_last=str(b[i])[2:]
    D[f"{i}"]=[]
    # print("oppo_last",oppo_last)
    for j in range(len(a)):
      oppo_last=str(a[j])[:2]
      # print("my_last",my_last)
      if(oppo_last==my_last):
        D[f"{i}"].append((i,j))
  print(D)
  longest_key = max(D, key=lambda k: len(D[k]))
  print(int(longest_key))
  return int(longest_key)

def minmax2(oppo_move,my_move):
  a=[]
  b=[]
  D={}
  for i in range(len(oppo_move)):
    a.append(str(oppo_move[i])[:])

  for j in range(len(my_move)):
    b.append(str(my_move[j])[:])
  # print(b)
  # print("a,b=",a,b)
  for i in range(len(b)):
    my_last=str(b[i])[2:]
    D[f"{i}"]=[]
    # print("oppo_last",oppo_last)
    for j in range(len(a)):
      oppo_last=str(a[j])[2:]
      # print("my_last",my_last)
      if(oppo_last==my_last):
        D[f"{i}"].append((i,j))

  longest_key = max(D, key=lambda k: len(D[k]))
  return int(longest_key)



```
